```python
from sqlalchemy import create_engine, text
import pymysql
import pandas as pd
import os
```

```python
# Secure SQLAlchemy connection: credentials are loaded from a local .env file.
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine

load_dotenv()
database_url = os.getenv("PIIT_DATABASE_URL")
if not database_url:
    raise RuntimeError(
        "PIIT_DATABASE_URL is not set. Copy .env.example to .env and add your local connection URL."
    )

engine1 = create_engine(database_url, pool_pre_ping=True)
print("Database engine configured; credentials are not displayed.")
```

```python
# function to run the query with parameters or query
def q(sql, **params):
    with engine1.connect() as conn:
        return pd.read_sql(text(sql), conn, params=params)
```

```python
emp = q("SELECT * FROM emp")
emp
```

```python
dept = q("SELECT * FROM dept")
dept
```

#### `1. List the names of analysts and salesmen.`

```python
# MYSQL Query
q("SELECT ename, job FROM emp WHERE job in('ANALYST','SALESMAN')")
```

```python
# Equivalent pandas code
result = emp[emp['job'].isin(['ANALYST', 'SALESMAN'])][['ename', 'job']]
result
```

#### `2. List details of employees who have joined before 30 Sep 81.`

```python
# MYSQL Query
q("SELECT * FROM emp where hiredate < '1981-09-30'")
```

```python
## Equivalent pandas code
emp['hiredate'] = pd.to_datetime(emp['hiredate']) # Convert to datetime first
result = emp.loc[emp['hiredate'] < '1981-09-30']
result
```

#### `3. List names of employees who are not managers.`

```python
# MYSQL Query
q("SELECT ename, job FROM emp WHERE job <> 'MANAGER'")
```

```python
## Equivalent pandas code
emp[emp['job'] != 'MANAGER'][['ename','job']]
```

#### `4. List the names of employees whose employee numbers are 7369, 7521, 7839, 7934, 7788.`

```python
# MYSQL Query
q("SELECT ename FROM emp WHERE empno IN (7369, 7521, 7839, 7934, 7788)")
```

```python
## Equivalent pandas code
emp[emp['empno'].isin([7369, 7521, 7839, 7934, 7788])] [['ename']]
```

#### `5. List employees not belonging to department 30, 40, or 10.`

```python
# MYSQL Query
q("SELECT ename FROM emp WHERE deptno NOT IN (30,40,10)")
```

```python
# Equivalent pandas code
emp[~emp['deptno'].isin([30,40,10])][['ename']]
```

#### 6.  `List employee names for those who have joined between 30 June and 31 Dec.'81`

```python
# MYSQL Query
q("SELECT ename FROM emp WHERE hiredate between '1981-06-30' and '1981-12-31'")
```

```python
# Equivalent pandas code
start_date = '1981-06-30'
end_date = '1981-12-31'
emp[emp['hiredate'].between(start_date, end_date)][['ename']]
```

#### 7.  `List the different designations in the company.`

```python
# MYSQL Query
q("SELECT DISTINCT(job) AS designation FROM emp")
```

```python
# Equivalent pandas code
emp[['job']].drop_duplicates().rename(columns={'job': 'designation'})                                                                     
```

#### 8.  `List the names of employees who are not eligible for commission.`

```python
# MYSQL Query
q("SELECT ename FROM emp WHERE comm IS NULL OR comm = 0")
```

```python
# Equivalent pandas code
emp[(emp['comm'].isna ()) | ( emp['comm'] == 0) ][['ename']] 
```

#### 9.  `List the name and designation of the employee who does not report to anybody.`

```python
# MYSQL Query
q("SELECT ename AS name, job AS designation FROM emp where mgr IS NULL")
```

```python
# Equivalent pandas code
emp[emp['mgr'].isna()][['ename','job']].rename(columns ={'ename':'name', 'job':'designation'})
```

#### '10. `List the employees not assigned to any department.`

```python
# MYSQL Query
q("SELECT ename FROM emp where deptno IS NULL")
```

```python
# Equivalent pandas code
emp[emp['deptno'].isna()][['ename']]
```

#### 11.  `List the employees who are eligible for commission.`

```python
# MYSQL Query
q("SELECT ename FROM emp WHERE comm IS NOT NULL AND comm != 0")
```

```python
# Equivalent pandas code
emp[emp['comm'].notna() & ( emp['comm'] != 0)][['ename']] 
```

#### 12.  `List employees whose names either start or end with “S”.`

```python
# MYSQL Query
q("SELECT ename FROM emp WHERE ename LIKE 'S%' OR ename LIKE '%S'")
```

```python
# Equivalent pandas code
emp[emp['ename'].str.contains(r'^S|S$', case=False, na=False)][['ename']]
```

#### 13.  `List names of employees whose names have “i” as the second character.`

```python
# MYSQL Query
q("SELECT ename FROM emp WHERE ename LIKE'_i%'")
```

```python
# Equivalent pandas code
emp[emp['ename'].str.contains(r'^.i', case=False, na=False)][['ename']] 
```

#### 14.  `List the number of employees working with the company.`

```python
# MYSQL Query
q("SELECT COUNT(empno) AS Number_of_Employees FROM emp")
```

```python
# Equivalent pandas code
#emp[['empno']].count().rename({'empno': 'Nunber_of_Employees'}) # same output as below
emp['empno'].agg(Nunber_of_Employees='count')
```

#### 15.  `List the number of designations available in the EMP table.`

```python
# MYSQL Query
q("SELECT COUNT(DISTINCT job) AS Number_of_Designations FROM emp")
```

```python
# Equivalent pandas code
#emp[['job']].count().rename({'job': 'Nunber_of_designations'})
pd.DataFrame({'Number_of_Designations': [emp['job'].nunique()]})
```

#### 16.  `List the total salaries paid to the employees.`

```python
# MYSQL Query
q("SELECT SUM(sal) AS Total_Salaries FROM emp")
```

```python
# Equivalent pandas code
#Total_Salaries = emp['sal'].sum()
#print(f'Total_Salaries: {Total_Salaries:.2f}')
emp['sal'].agg(Total_Salaries='sum')
```

#### 17.  `List the maximum, minimum and average salary in the company.`

```python
# MYSQL Query
q("SELECT MAX(sal) AS Maximum_Salary, MIN(sal) AS Minimum_Salary, AVG(Sal) AS Average_salary FROM emp")
```

```python
# Equivalent pandas code
emp["sal"].agg(
    Maximum_Salary="max", Minimum_Salary="min", Average_salary="mean"
)
```

#### 18.  `List the maximum salary paid to a salesman`

```python
# MYSQL Query
q("SELECT MAX(sal) FROM emp where job = 'salesman'")
```

```python
# Equivalent pandas code
emp[emp['job'] == 'SALESMAN']['sal'].agg(Maximum_Salary='max')
```
